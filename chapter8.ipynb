# Chapter 8: Wrapping Up

Things to consider before implementing machine learning algorithms.
- Do you want to do exploratory analysis and just see if you find something interesting in the data?
- Do you already have a particular goal in mind?
- Do you have the right data to evaluate an algorithm?
- Do you know how to measure your prediction to see if its actually working?

Keep in mind that model building is often part of a feedback circle of collecting new data, cleaning data, building models, and analyzing the models. Analyzing the mistakes a model makes can often be informative about what is missing in the data, what additional data could be collected, or how the task could be reformulated to make machine learning more effective.
